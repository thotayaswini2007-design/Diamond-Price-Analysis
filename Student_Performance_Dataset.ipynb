```python
url="https://raw.githubusercontent.com/ageron/data/master/student-mat.csv"
df=pd.read_csv(url,sep=";")
```

```python
print("G3 Final Grade Stastistics")
print("Mean:",round(df['G3'].mean(),2))
print("Median:",df['G3'].median())
print("Mode:",df['G3'].mode()[0])
print("Variance:",round(df['G3'].var(),2))
print("Standard Deviation:",round(df['G3'].std(),2))
print("/n all numeric columns")
print(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,4))
sns.boxplot(x=df['G3'])
plt.title('G3 Final Grade - outliers')
plt.show()
```

```python
print(df.corr(numeric_only=True)['G3'].sort_values(ascending=False))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap='coolwarm')
plt.show()
sns.scatterplot(x='studytime',y='G3',data=df)
plt.title('Study Time vs G3')
plt.show()
```

```python
sns.countplot(x='studytime',data=df)
plt.title('Bar CHart:Study time distribution')
plt.show()
```

```python
sns.histplot(df['G3'],bins=10)
plt.title('Histogram:Final grade distribution')
plt.show()
```

```python
sns.scatterplot(x='studytime',y='G3',data=df)
plt.title("Sactter plot:study time vs G3")
plt.show()
```

```python
df.groupby('studytime')['G3'].mean().plot(kind='line',marker='o')
plt.title('Line chart: avg G3 vs study time')
plt.show()
```
