```python
from sklearn.preprocessing import LabelEncoder
encoder =LabelEncoder()
for columns in df.columns:
  if df[columns].dtype=="object":
    df[columns]=encoder.fit_transform(df[columns])
```

```python
correlation=df.corr()
print(correlation)
```

```python
x=df.drop("Attrition",axis=1)
y=df["Attrition"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lr_model = LogisticRegression(max_iter =1000)
lr_model.fit(x_train, y_train)
```

```python
x_pred_lr =lr_model.predict(x_test)
print(x_pred_lr)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score=accuracy_score(y_test,x_pred_lr)
print(accuracy_score)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix=confusion_matrix(y_test,x_pred_lr)
print(confusion_matrix)
```

```python
from sklearn.metrics import classification_report
lr_classification_report=classification_report(y_test,x_pred_lr)
print(lr_classification_report)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model=KNeighborsClassifier(n_neighbors=5)
knn_model.fit(x_train,y_train)
```

```python
x_pred_knn=knn_model.predict(x_test)
print(x_pred_knn)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
print('Accuracy:',accuracy_score(y_test,x_pred_knn))
print(confusion_matrix(y_test,x_pred_knn))
print(classification_report(y_test,x_pred_knn))
```

```python
from sklearn.tree import DecisionTreeClassifier
dtree_model = DecisionTreeClassifier()
dtree_model.fit(x_train, y_train)
```

```python
x_pred_dtree = dtree_model.predict(x_test)
print(x_pred_dtree)
```

```python
print('Accuracy:', accuracy_score(y_test, x_pred_dtree))
print(confusion_matrix(y_test, x_pred_dtree))
print(classification_report(y_test, x_pred_dtree))
```
